```python
from langchain.chat_models import init_chat_model
```

```python
model=init_chat_model("gpt-4o-mini")
model
```

```python
for chunk in model.stream("what is capital of india?"):
    print(chunk.text(),end="",flush=True)
```

```python
from langchain_openai import ChatOpenAI
```

```python
from langchain_core.tools import tool
@tool 
def get_weather(location: str) -> str:
    """Get the current weather for a given location."""
    # This is a mock implementation. In a real implementation, you would call a weather API.
    return f"The current weather in {location} is sunny with a temperature of 25°C."    
model_with_tools =  model.bind_tools([get_weather])
response = model_with_tools.invoke("What is the weather in New York?")
print(response)
```

```python
response.tool_calls
```

```python
from langchain.agents import create_tool_calling_agent, AgentExecutor
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that provides weather information."),
    MessagesPlaceholder("chat_history", optional=True),
    ("human", "{input}"),
    MessagesPlaceholder("agent_scratchpad"),
])

agent = create_tool_calling_agent(
    llm=model,
    tools=[get_weather],
    prompt=prompt,
)
agent_executor = AgentExecutor(agent=agent, tools=[get_weather], verbose=True)
agent_executor
```
