```python
from typing import TypedDict
from langgraph.graph import StateGraph  # noqa: F401
```

```python
class AgentState(TypedDict):
    name: str
    age: int
    final: str
```

```python
def first_node(state:AgentState)->AgentState:
    """First node that processes the input and updates the state."""
    state["final"] = f"Hello {state['name']}."
    return state
```

```python
def second_node(state:AgentState)->AgentState:
    """Second node that processes the input and updates the state."""
    state["final"] += f" You are {state['age']} years old."
    return state
```

```python
graph=StateGraph(AgentState)
graph.add_node("FirstNode",first_node)
graph.add_node("SecondNode",second_node)
graph.add_edge("FirstNode","SecondNode")
graph.set_entry_point("FirstNode")
graph.set_finish_point("SecondNode")
app=graph.compile()
```

```python
app
```

```python
output = app.invoke({"name":"jacksparrow","age":30})
print(output)
```
